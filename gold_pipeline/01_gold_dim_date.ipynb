# dim_date - static calendar

One row per day. Loaded once for several years ahead; the fact quality gate fails if a fact date is outside the calendar.

```python
spark.sql(f"""
INSERT OVERWRITE {GOLD}.dim_date
    (date_key, full_date, year, quarter, month, month_name, day_of_month, day_name, is_weekend)
SELECT
    CAST(date_format(d, 'yyyyMMdd') AS INT),
    d,
    year(d),
    quarter(d),
    month(d),
    date_format(d, 'MMMM'),
    dayofmonth(d),
    date_format(d, 'EEEE'),
    dayofweek(d) IN (1, 7)
FROM (
    SELECT explode(sequence(DATE '{START_DATE}', DATE '{END_DATE}', INTERVAL 1 DAY)) AS d
)
""")
```
