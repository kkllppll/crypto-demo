## Silver Table Verification

```python
dbutils.widgets.text("catalog", "dbr_dev_ua5816bd")
dbutils.widgets.text("silver_schema", "crypto_silver")
```

```python
from pyspark.sql import functions as F

catalog = dbutils.widgets.get("catalog")
silver_schema = dbutils.widgets.get("silver_schema")
silver_prices = f"{catalog}.{silver_schema}.prices"
```

```python
spark.table(silver_prices).printSchema()
spark.table(silver_prices).orderBy(F.col("event_time").desc()).limit(5).show(truncate=False)
```
