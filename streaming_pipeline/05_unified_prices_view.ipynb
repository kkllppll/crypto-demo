## Unified prices view — for the dashboard
`ohlc_prices` (Batch) and `ticks_prices` (Streaming) have different shapes (candles vs last-trade ticks), so they stay as separate tables rather than being forced into one. This view projects both down to a common shape the dashboard can query as a single source.

```python
dbutils.widgets.text("catalog", "dbr_dev_ua5816bd")
dbutils.widgets.text("silver_schema", "team_crypto_silver")
```

```python
catalog = dbutils.widgets.get("catalog")
silver_schema = dbutils.widgets.get("silver_schema")
view_name = f"{catalog}.{silver_schema}.prices_unified"
ohlc_table = f"{catalog}.{silver_schema}.ohlc_prices"
ticks_table = f"{catalog}.{silver_schema}.ticks_prices"
```

```python
spark.sql(f"""
CREATE OR REPLACE VIEW {view_name} AS
SELECT symbol, close AS price, event_time, source FROM {ohlc_table}
UNION ALL
SELECT symbol, price_usd AS price, event_time, source FROM {ticks_table}
""")
```
